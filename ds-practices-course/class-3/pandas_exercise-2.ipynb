```python

ipl_df['tosswinner'].unique()
```

```python

ipl_df['winningteam'].unique()
```

```python
ipl_df['winningteam'] = ipl_df['winningteam'].replace(cleanup_map)
```

```python
bool_map = {
    'N': False,
    'Y': True
}

ipl_df['superover'] = ipl_df['superover'].map(bool_map).fillna(False)
```
