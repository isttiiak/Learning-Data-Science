# Numpy array creation and methods

```python
import numpy as np

# create a basic 1D NumPy array
arr = np.array([1, 2, 3, 4, 5])

print(arr)
```

```python
default_zero_array = np.zeros(5) # for 1d

default_zero_array
```

```python
default_zero_array = np.zeros((3,4)) # for 2d --> remember shape need to define within parenthesis. First one represent row and second one represent column.

default_zero_array
```

```python
default_zero_array = np.ones(5) # for 1d

default_zero_array
```

```python
default_zero_array = np.full(5, 2) # for 1d. first one for shape then the custom filler value.

default_zero_array
```

```python
default_zero_array = np.full((4, 3), 2) # for 1d. first one for shape then the custom filler value.

default_zero_array
```
