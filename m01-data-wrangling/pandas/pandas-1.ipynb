# Creating a DataFrame from a dictionary and saving it as csv, excel, and json files

To create a DataFrame from a dictionary, you can use the `DataFrame` constructor from pandas. And we can save it as csv, excel, and json files using the `to_csv`, `to_excel`, and `to_json` methods respectively.
Here is an example of how to create a DataFrame from a dictionary and save it as different file formats:

```
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}```

```df = pd.DataFrame(data)

df.to_csv('file.csv', index=False)

df.to_excel('file.xlsx', index=False)

df.to_json('file.json', orient='records')```

In this example, we first create a dictionary with some sample data. Then we convert it into a DataFrame using ```pd.DataFrame()```. Finally, we save the DataFrame to a CSV file, an Excel file, and a JSON file using the respective methods. The ```index=False``` parameter is used to prevent pandas from writing row indices to the files.

````python
# create dataframe and export it as csv, excel and json

data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}

custom_df = pd.DataFrame(data)

custom_df
# here we can see row index number by default
````

````python
# removing the index while exporting

# to csv
custom_df.to_csv("custom_data.csv", index=False)
````

````python
# to excel
custom_df.to_excel("custom_data.xlsx", index=False)
````

````python
# to json
custom_df.to_json("custom_data.json", index=False)
````

# Modifying and advanced filtering DataFrame in pandas

* Add a column to a DataFrame: You can add a new column to a DataFrame by assigning a value to a new column name. For example, `df['New_Column'] = value` will add a new column named 'New_Column' with the specified value. This value can be a constant, a list, or a result of an operation on existing columns. The new column will be added at the end of the DataFrame.
* Another way is inserting a column at a specific position using the `insert()` method. For example, `df.insert(loc=1, column='New_Column', value=value)` will insert a new column named 'New_Column' at index 1 (second position) with the specified value.





````python
df["DISCOUNT"] = df["PRICEEACH"] * 0.90

df
````

````python
df[["PRICEEACH", "DISCOUNT"]]
````

````python
# single column reorder using pop and insert

col = df.pop("DISCOUNT")

df.insert(3, "DISCOUNT", col)

df
````

# Handling missing values in pandas

To find missing values in a DataFrame, you can use the `isnull()` method, which returns a DataFrame of the same shape with boolean values indicating whether each element is missing (True) or not (False). To get the count of missing values in each column, you can chain the `sum()` method to `isnull()`, like this: `df.isnull().sum()`. This will return a Series with the count of missing values for each column in the DataFrame.

To handle missing values, you have several options:
1. **Drop missing values**: You can use the `dropna()` method to remove rows or columns that contain missing values. For example, `df.dropna()` will drop all rows that have at least one missing value, while `df.dropna(axis=1)` will drop all columns that have at least one missing value.

2. **Fill missing values**: You can use the `fillna()` method to replace missing values with a specified value or a method. For example, `df.fillna(0)` will replace all missing values with 0, while `df.fillna(method='ffill')` will forward-fill the missing values using the last valid observation.

3. **Interpolate missing values**: You can use the `interpolate()` method to estimate missing values based on the existing data. For example, `df.interpolate(method='linear')` will perform linear interpolation to fill in the missing values.

4. **Replace missing values with the mean, median, or mode**: You can calculate the mean, median, or mode of a column and use it to fill in the missing values. For example, `df['Column_Name'].fillna(df['Column_Name'].mean(), inplace=True)` will replace missing values in 'Column_Name' with the mean of that column.

5. **Use machine learning algorithms**: For more complex datasets, you can use machine learning algorithms to predict and fill in missing values based on the patterns in the data. This approach is more advanced and may require additional libraries such as scikit-learn.
