```python
# ── 4. Full missing value report ──────────────────────────────

missing = pd.DataFrame({
    'missing_count': df.isnull().sum(),
    'missing_%': (df.isnull().sum() / len(df) * 100).round(2)
})

print(f"Missing Data with percentage:")
print(missing[missing['missing_count'] > 0])
```

```python
# different way to calculate percentage

miss_per = (df.isnull().mean() * 100).round(2)

miss_per[miss_per > 0]
```
