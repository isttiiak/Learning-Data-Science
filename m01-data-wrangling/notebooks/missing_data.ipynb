```python
import pandas as pd
import numpy as np
# Visual heatmap of missing data (great for EDA notebooks)
import seaborn as sns
import matplotlib.pyplot as plt
```

```python


# Realistic employee dataset with missing values
df = pd.DataFrame({
    'emp_id':   [101, 102, 103, 104, 105, 106],
    'name':     ['Istiak', 'Rafi', 'Nadia', 'Sadia', 'Mahadi', 'Aydan'],
    'age':      [28, 35, np.nan, 42, np.nan, 30],
    'salary':   [55000, 72000, 68000, np.nan, np.nan, 61000],
    'dept':     ['Engineering', 'HR', 'Engineering', np.nan, 'HR', 'Engineering'],
    'joined':   ['2020-01', '2018-05', '2019-11', '2021-03', np.nan, '2022-07']
})
```

```python
# --- Basic detection ---
print(df.isnull())              # True/False for every cell
```

```python
print(df.isnull().sum())        # count of nulls per column
# age      2
# salary   2
# dept     1
# joined   1
```

```python
print(df.isnull().mean() * 100) # percentage missing per column
# age      33.3%
# salary   33.3%
```

```python
# Which rows have ANY missing value?
print(df[df.isnull().any(axis=1)])
```

```python
# Check if missing salary correlates with another column
# If salary is MNAR — high earners skip it — mean imputation will underestimate
print(df.groupby(df['salary'].isnull())['age'].mean())
# If avg age is very different between missing/not-missing salary rows → MAR or MNAR
```
